```python
# ============================================================
# CELL 4: Load Stable Diffusion 1.5 - FP32
# ============================================================

import os
import torch
from diffusers import StableDiffusionPipeline

# Model configuration
MODEL_ID = "stable-diffusion-v1-5/stable-diffusion-v1-5"

DEVICE = "cuda:0"
BASELINE_DTYPE = torch.float32

# Reproducibility
SEED = 12345

# Output directories
PROJECT_DIR = "/kaggle/working/M1_SD15"

IMAGE_DIR = os.path.join(PROJECT_DIR, "images")
METRICS_DIR = os.path.join(PROJECT_DIR, "metrics")
METADATA_DIR = os.path.join(PROJECT_DIR, "metadata")
MODEL_OUTPUT_DIR = os.path.join(PROJECT_DIR, "model_info")

for directory in [
    IMAGE_DIR,
    METRICS_DIR,
    METADATA_DIR,
    MODEL_OUTPUT_DIR
]:
    os.makedirs(directory, exist_ok=True)

print("=" * 65)
print("LOADING STABLE DIFFUSION 1.5")
print("=" * 65)

print("\nModel:", MODEL_ID)
print("Device:", DEVICE)
print("Precision:", BASELINE_DTYPE)
print("Seed:", SEED)

# Clear unused CUDA memory
torch.cuda.empty_cache()

# Load original pretrained checkpoint
pipe = StableDiffusionPipeline.from_pretrained(
    MODEL_ID,
    torch_dtype=BASELINE_DTYPE,
    use_safetensors=True
)

# Move pipeline to GPU 0
pipe = pipe.to(DEVICE)

# Do not change the original scheduler yet.
print("\nORIGINAL CHECKPOINT SCHEDULER")
print("Scheduler:", pipe.scheduler.__class__.__name__)
print("Prediction type:",
      pipe.scheduler.config.prediction_type)

print("\nScheduler configuration:")
print(dict(pipe.scheduler.config))

# GPU memory after model loading
torch.cuda.synchronize()

allocated = torch.cuda.memory_allocated(0) / (1024 ** 2)
reserved = torch.cuda.memory_reserved(0) / (1024 ** 2)

print("\nGPU MEMORY AFTER MODEL LOADING")
print("Allocated:", round(allocated, 2), "MB")
print("Reserved:", round(reserved, 2), "MB")

print("\nMODEL LOADED SUCCESSFULLY")
print("=" * 65)
```

```python
# ============================================================
# CELL 7: OFFICIAL BASELINE - SD 1.5, FP32, PNDM, 30 STEPS
# ============================================================

import os
import time
import torch
import numpy as np
import pandas as pd
import cv2
from PIL import Image

# ------------------------------------------------------------
# 1. COMMON TEAM PROMPTS
# ------------------------------------------------------------

PROMPTS = [
    "A red sports car parked beside a yellow bicycle under a tree, with a mountain in the background.",
    "A wooden table with a blue ceramic vase, three red apples and an open book, photographed in natural light.",
    "A small white dog sitting on a green sofa, with a brown cat sleeping beside it.",
    "A futuristic city with flying vehicles, glass skyscrapers and people walking on a wet street at night.",
    "A woman wearing a red jacket standing to the left of a blue motorcycle, with a lake behind them.",
    "A wooden cabin surrounded by pine trees, snow-covered mountains in the background and a river flowing in front.",
    "A black laptop placed on a wooden desk, with a coffee cup on its right and a notebook on its left.",
    "A realistic garden containing yellow sunflowers, purple tulips and a small stone fountain.",
    "A modern living room with a grey sofa, a round coffee table and a large window overlooking a city.",
    "A golden retriever running on a beach while two birds fly above the ocean.",
    "A red umbrella leaning against a blue chair, both placed under a large tree in a quiet park.",
    "A glass bottle filled with orange juice placed behind a white plate containing two green pears and one sliced orange."
]

# ------------------------------------------------------------
# 2. BASELINE CONFIGURATION
# ------------------------------------------------------------

BASELINE_STEPS = 30
BASELINE_GUIDANCE = 7.5
BASELINE_HEIGHT = 512
BASELINE_WIDTH = 512
BASELINE_SEED = 12345

BASELINE_SCHEDULER = pipe.scheduler.__class__.__name__
BASELINE_RUN_ID = "baseline_pndm_fp32_30_512"

baseline_rows = []

print("=" * 65)
print("OFFICIAL BASELINE BENCHMARK")
print("=" * 65)
print(f"Model: {MODEL_ID}")
print(f"Scheduler: {BASELINE_SCHEDULER}")
print("Precision: FP32")
print(f"Steps: {BASELINE_STEPS}")
print(f"Resolution: {BASELINE_WIDTH} x {BASELINE_HEIGHT}")
print(f"Guidance scale: {BASELINE_GUIDANCE}")
print(f"Seed: {BASELINE_SEED}")
print(f"Number of prompts: {len(PROMPTS)}")
print("=" * 65)

# ------------------------------------------------------------
# 3. GENERATE AND MEASURE EACH PROMPT
# ------------------------------------------------------------

for prompt_id, prompt in enumerate(PROMPTS, start=1):

    print(f"\nGenerating prompt {prompt_id}/{len(PROMPTS)}")

    generator = torch.Generator(
        device="cuda:0"
    ).manual_seed(BASELINE_SEED)

    # Reset memory counters for this individual generation
    torch.cuda.empty_cache()
    torch.cuda.reset_peak_memory_stats(0)
    torch.cuda.synchronize(0)

    start_time = time.perf_counter()

    with torch.inference_mode():
        result = pipe(
            prompt=prompt,
            height=BASELINE_HEIGHT,
            width=BASELINE_WIDTH,
            num_inference_steps=BASELINE_STEPS,
            guidance_scale=BASELINE_GUIDANCE,
            generator=generator
        )

    torch.cuda.synchronize(0)
    latency = time.perf_counter() - start_time

    # Capture GPU memory before any other GPU operation
    peak_allocated = torch.cuda.max_memory_allocated(0) / (1024 ** 2)
    peak_reserved = torch.cuda.max_memory_reserved(0) / (1024 ** 2)

    # --------------------------------------------------------
    # 4. SAVE GENERATED IMAGE
    # --------------------------------------------------------

    image = result.images[0]

    image_filename = f"baseline_prompt_{prompt_id:02d}.png"
    image_path = os.path.join(IMAGE_DIR, image_filename)

    image.save(image_path)

    # --------------------------------------------------------
    # 5. BASIC IMAGE QUALITY STATISTICS
    # --------------------------------------------------------

    image_array = np.array(image.convert("RGB"))

    gray_image = cv2.cvtColor(
        image_array,
        cv2.COLOR_RGB2GRAY
    )

    sharpness = cv2.Laplacian(
        gray_image,
        cv2.CV_64F
    ).var()

    brightness = np.mean(gray_image)
    contrast = np.std(gray_image)

    # --------------------------------------------------------
    # 6. STORE METRICS
    # --------------------------------------------------------

    row = {
        "member": "Member 1",
        "model_id": MODEL_ID,
        "run_id": BASELINE_RUN_ID,
        "precision": "FP32",
        "scheduler": BASELINE_SCHEDULER,
        "steps": BASELINE_STEPS,
        "guidance_scale": BASELINE_GUIDANCE,
        "height": BASELINE_HEIGHT,
        "width": BASELINE_WIDTH,
        "seed": BASELINE_SEED,
        "prompt_id": prompt_id,
        "latency_seconds": round(latency, 4),
        "peak_allocated_mb": round(peak_allocated, 2),
        "peak_reserved_mb": round(peak_reserved, 2),
        "model_directory_size_mb": None,
        "clip_similarity": None,
        "sharpness_laplacian_var": round(float(sharpness), 4),
        "brightness_mean": round(float(brightness), 4),
        "contrast_std": round(float(contrast), 4),
        "notes": "Official FP32 baseline; CLIP evaluation pending"
    }

    baseline_rows.append(row)

    print(f"  Latency: {latency:.2f} seconds")
    print(f"  Peak allocated: {peak_allocated:.2f} MB")
    print(f"  Peak reserved: {peak_reserved:.2f} MB")
    print(f"  Image saved: {image_filename}")

    del result, image, image_array, gray_image, generator

# ------------------------------------------------------------
# 7. SAVE BASELINE METRICS
# ------------------------------------------------------------

baseline_df = pd.DataFrame(baseline_rows)

baseline_csv_path = os.path.join(
    METRICS_DIR,
    "baseline_fp32_pndm.csv"
)

baseline_df.to_csv(baseline_csv_path, index=False)

print("\n" + "=" * 65)
print("BASELINE BENCHMARK COMPLETED")
print("=" * 65)

print(f"Images generated: {len(baseline_rows)}")
print(f"Metrics saved: {baseline_csv_path}")

display(
    baseline_df[
        [
            "prompt_id",
            "latency_seconds",
            "peak_allocated_mb",
            "peak_reserved_mb",
            "sharpness_laplacian_var",
            "brightness_mean",
            "contrast_std"
        ]
    ]
)

print("\nAverage generation latency:",
      round(baseline_df["latency_seconds"].mean(), 2), "seconds")

print("Average peak allocated memory:",
      round(baseline_df["peak_allocated_mb"].mean(), 2), "MB")

print("Average peak reserved memory:",
      round(baseline_df["peak_reserved_mb"].mean(), 2), "MB")
```

```python
# ============================================================
# CELL 19: CLIP SEMANTIC SIMILARITY EVALUATION
# ============================================================

import os
import torch
import pandas as pd
import numpy as np

from PIL import Image
from transformers import CLIPModel, CLIPProcessor

CLIP_MODEL_ID = "openai/clip-vit-base-patch32"

print("Loading CLIP model on CPU...")

clip_processor = CLIPProcessor.from_pretrained(CLIP_MODEL_ID)

clip_model = CLIPModel.from_pretrained(
    CLIP_MODEL_ID
).to("cpu")

clip_model.eval()

print("CLIP model loaded successfully.")

# ------------------------------------------------------------
# CONFIGURATIONS TO EVALUATE
# ------------------------------------------------------------

configurations = {
    "baseline_fp32_pndm_30_512": {
        "folder": IMAGE_DIR,
        "pattern": "baseline_prompt_{:02d}.png"
    },
    "fp16_pndm_30_512": {
        "folder": IMAGE_DIR,
        "pattern": "fp16_prompt_{:02d}.png"
    },
    "fp16_ddim_30_512": {
        "folder": IMAGE_DIR,
        "pattern": "fp16_ddim_30_512_prompt_{:02d}.png"
    },
    "fp16_pndm_30_384": {
        "folder": IMAGE_DIR,
        "pattern": "fp16_pndm_30_384_prompt_{:02d}.png"
    },
    "combined_fp16_pndm_10_384": {
        "folder": IMAGE_DIR,
        "pattern": "combined_fp16_pndm_10_384_prompt_{:02d}.png"
    }
}

# Prompt 10 in the combined experiment was safety-filtered.
safety_filtered = {
    ("combined_fp16_pndm_10_384", 10)
}

clip_rows = []

# ------------------------------------------------------------
# EVALUATE EACH IMAGE
# ------------------------------------------------------------

for config_name, config in configurations.items():

    print(f"\nEvaluating: {config_name}")

    for prompt_id, prompt in enumerate(PROMPTS, start=1):

        image_path = os.path.join(
            config["folder"],
            config["pattern"].format(prompt_id)
        )

        if (config_name, prompt_id) in safety_filtered:
            print(f"Prompt {prompt_id:02d}: SKIPPED - safety filtered")

            clip_rows.append({
                "configuration": config_name,
                "prompt_id": prompt_id,
                "image_path": image_path,
                "clip_similarity": np.nan,
                "status": "safety_filtered"
            })
            continue

        if not os.path.exists(image_path):
            print(f"Prompt {prompt_id:02d}: IMAGE NOT FOUND")

            clip_rows.append({
                "configuration": config_name,
                "prompt_id": prompt_id,
                "image_path": image_path,
                "clip_similarity": np.nan,
                "status": "missing_image"
            })
            continue

        image = Image.open(image_path).convert("RGB")

        inputs = clip_processor(
            text=[prompt],
            images=[image],
            return_tensors="pt",
            padding=True
        )

        with torch.inference_mode():
            outputs = clip_model(**inputs)

            image_features = outputs.image_embeds
            text_features = outputs.text_embeds

            image_features = torch.nn.functional.normalize(
                image_features,
                p=2,
                dim=-1
            )

            text_features = torch.nn.functional.normalize(
                text_features,
                p=2,
                dim=-1
            )

            similarity = torch.sum(
                image_features * text_features,
                dim=-1
            ).item()

        clip_rows.append({
            "configuration": config_name,
            "prompt_id": prompt_id,
            "image_path": image_path,
            "clip_similarity": similarity,
            "status": "evaluated"
        })

        print(
            f"Prompt {prompt_id:02d}: "
            f"CLIP similarity = {similarity:.4f}"
        )

# ------------------------------------------------------------
# SAVE DETAILED RESULTS
# ------------------------------------------------------------

clip_df = pd.DataFrame(clip_rows)

clip_output = os.path.join(
    METRICS_DIR,
    "clip_semantic_similarity.csv"
)

clip_df.to_csv(clip_output, index=False)

# ------------------------------------------------------------
# SUMMARY
# ------------------------------------------------------------

clip_summary = (
    clip_df[clip_df["status"] == "evaluated"]
    .groupby("configuration")
    .agg(
        evaluated_images=("clip_similarity", "count"),
        average_clip_similarity=("clip_similarity", "mean"),
        minimum_clip_similarity=("clip_similarity", "min"),
        maximum_clip_similarity=("clip_similarity", "max"),
        std_clip_similarity=("clip_similarity", "std")
    )
    .reset_index()
)

summary_output = os.path.join(
    METRICS_DIR,
    "clip_semantic_summary.csv"
)

clip_summary.to_csv(summary_output, index=False)

print("\n" + "=" * 65)
print("CLIP SEMANTIC EVALUATION SUMMARY")
print("=" * 65)

print(clip_summary.to_string(index=False))

print("\nDetailed results:", clip_output)
print("Summary results:", summary_output)

print("\nImage status counts:")
print(clip_df["status"].value_counts())
```

```python
# ============================================================
# CELL 22: GENERATE VISUAL COMPARISON CONTACT SHEETS
# ============================================================

import os
from PIL import Image, ImageDraw
import matplotlib.pyplot as plt

configurations = {
    "FP32 PNDM 30 / 512": "baseline_prompt_{:02d}.png",
    "FP16 PNDM 30 / 512": "fp16_prompt_{:02d}.png",
    "FP16 DDIM 30 / 512": "fp16_ddim_30_512_prompt_{:02d}.png",
    "FP16 PNDM 30 / 384": "fp16_pndm_30_384_prompt_{:02d}.png",
    "Combined 10 / 384": "combined_fp16_pndm_10_384_prompt_{:02d}.png"
}

SHEET_DIR = os.path.join(PROJECT_DIR, "contact_sheets")
os.makedirs(SHEET_DIR, exist_ok=True)

THUMB_SIZE = (192, 192)
LABEL_HEIGHT = 25
COLS = 4
ROWS = 3

for config_name, filename_pattern in configurations.items():

    sheet_width = COLS * THUMB_SIZE[0]
    sheet_height = ROWS * (THUMB_SIZE[1] + LABEL_HEIGHT)

    sheet = Image.new(
        "RGB",
        (sheet_width, sheet_height),
        "white"
    )

    draw = ImageDraw.Draw(sheet)

    for idx in range(1, 13):

        filename = filename_pattern.format(idx)
        image_path = os.path.join(IMAGE_DIR, filename)

        if not os.path.exists(image_path):
            print(f"Missing: {filename}")
            continue

        image = Image.open(image_path).convert("RGB")
        image = image.resize(THUMB_SIZE)

        row = (idx - 1) // COLS
        col = (idx - 1) % COLS

        x = col * THUMB_SIZE[0]
        y = row * (THUMB_SIZE[1] + LABEL_HEIGHT)

        sheet.paste(image, (x, y))
        draw.text(
            (x + 5, y + THUMB_SIZE[1] + 4),
            f"Prompt {idx:02d}",
            fill="black"
        )

    safe_name = config_name.replace(" ", "_").replace("/", "_")
    output_path = os.path.join(
        SHEET_DIR,
        f"{safe_name}.png"
    )

    sheet.save(output_path)

    print(f"Saved: {output_path}")

    plt.figure(figsize=(14, 10))
    plt.imshow(sheet)
    plt.title(config_name)
    plt.axis("off")
    plt.show()

print("\nAll contact sheets generated.")
```

```python
# ============================================================
# CELL 23: CONSOLIDATE ALL EXPERIMENT RESULTS
# ============================================================

import os
import pandas as pd
import numpy as np

# ------------------------------------------------------------
# 1. EXPERIMENT CSV FILES
# ------------------------------------------------------------

experiment_files = [
    "baseline_fp32_pndm.csv",
    "fp16_pndm_30_512.csv",
    "fp16_ddim_30_512.csv",
    "fp16_pndm_step_optimization.csv",
    "fp16_ddim_step_optimization.csv",
    "fp16_pndm_30_384.csv",
    "combined_fp16_pndm_10_384.csv"
]

all_results = []

print("=" * 65)
print("LOADING EXPERIMENT RESULTS")
print("=" * 65)

for filename in experiment_files:

    file_path = os.path.join(METRICS_DIR, filename)

    if os.path.exists(file_path):

        df = pd.read_csv(file_path)
        all_results.append(df)

        print(f"Loaded: {filename} | Rows: {len(df)}")

    else:
        print(f"NOT FOUND: {filename}")

if not all_results:
    raise FileNotFoundError("No experiment CSV files were found.")

# ------------------------------------------------------------
# 2. COMBINE EXPERIMENTS
# ------------------------------------------------------------

master_df = pd.concat(
    all_results,
    ignore_index=True
)

# ------------------------------------------------------------
# 3. LOAD CLIP RESULTS
# ------------------------------------------------------------

clip_path = os.path.join(
    METRICS_DIR,
    "clip_semantic_similarity.csv"
)

clip_df = pd.read_csv(clip_path)

# Map CLIP configuration names to experiment parameters
clip_mapping = {
    "baseline_fp32_pndm_30_512": {
        "precision": "FP32",
        "scheduler": "PNDM",
        "steps": 30,
        "height": 512,
        "width": 512
    },
    "fp16_pndm_30_512": {
        "precision": "FP16",
        "scheduler": "PNDM",
        "steps": 30,
        "height": 512,
        "width": 512
    },
    "fp16_ddim_30_512": {
        "precision": "FP16",
        "scheduler": "DDIM",
        "steps": 30,
        "height": 512,
        "width": 512
    },
    "fp16_pndm_30_384": {
        "precision": "FP16",
        "scheduler": "PNDM",
        "steps": 30,
        "height": 384,
        "width": 384
    },
    "combined_fp16_pndm_10_384": {
        "precision": "FP16",
        "scheduler": "PNDM",
        "steps": 10,
        "height": 384,
        "width": 384
    }
}

clip_df["precision"] = clip_df["configuration"].map(
    lambda x: clip_mapping.get(x, {}).get("precision")
)

clip_df["scheduler"] = clip_df["configuration"].map(
    lambda x: clip_mapping.get(x, {}).get("scheduler")
)

clip_df["steps"] = clip_df["configuration"].map(
    lambda x: clip_mapping.get(x, {}).get("steps")
)

clip_df["height"] = clip_df["configuration"].map(
    lambda x: clip_mapping.get(x, {}).get("height")
)

clip_df["width"] = clip_df["configuration"].map(
    lambda x: clip_mapping.get(x, {}).get("width")
)

clip_df = clip_df.rename(
    columns={"status": "clip_status"}
)

# ------------------------------------------------------------
# 4. MERGE CLIP SCORES
# ------------------------------------------------------------

merge_columns = [
    "precision",
    "scheduler",
    "steps",
    "height",
    "width",
    "prompt_id"
]

clip_merge = clip_df[
    merge_columns + ["clip_similarity", "clip_status"]
].copy()

master_df = master_df.drop(
    columns=["clip_similarity"],
    errors="ignore"
)

master_df = master_df.merge(
    clip_merge,
    on=merge_columns,
    how="left"
)

# ------------------------------------------------------------
# 5. ADD CONFIGURATION IDENTIFIERS
# ------------------------------------------------------------

def identify_configuration(row):

    if (
        row["precision"] == "FP32"
        and row["scheduler"] == "PNDM"
        and row["steps"] == 30
        and row["height"] == 512
    ):
        return "FP32_PNDM_30_512"

    if (
        row["precision"] == "FP16"
        and row["scheduler"] == "PNDM"
        and row["steps"] == 30
        and row["height"] == 512
    ):
        return "FP16_PNDM_30_512"

    if (
        row["precision"] == "FP16"
        and row["scheduler"] == "DDIM"
        and row["steps"] == 30
        and row["height"] == 512
    ):
        return "FP16_DDIM_30_512"

    if (
        row["precision"] == "FP16"
        and row["scheduler"] == "PNDM"
        and row["steps"] == 30
        and row["height"] == 384
    ):
        return "FP16_PNDM_30_384"

    if (
        row["precision"] == "FP16"
        and row["scheduler"] == "PNDM"
        and row["steps"] == 10
        and row["height"] == 384
    ):
        return "COMBINED_FP16_PNDM_10_384"

    return (
        f"{row['precision']}_{row['scheduler']}_"
        f"{row['steps']}_{row['height']}"
    )

master_df["configuration"] = master_df.apply(
    identify_configuration,
    axis=1
)

# ------------------------------------------------------------
# 6. SAVE MASTER DATASET
# ------------------------------------------------------------

master_df = master_df.sort_values(
    by=["configuration", "prompt_id"]
).reset_index(drop=True)

master_output = os.path.join(
    METRICS_DIR,
    "M1_SD15_MASTER_RESULTS.csv"
)

master_df.to_csv(master_output, index=False)

# ------------------------------------------------------------
# 7. DISPLAY SUMMARY
# ------------------------------------------------------------

summary = master_df.groupby(
    "configuration"
).agg(
    image_count=("prompt_id", "count"),
    average_latency=("latency_seconds", "mean"),
    average_allocated_vram=("peak_allocated_mb", "mean"),
    average_reserved_vram=("peak_reserved_mb", "mean"),
    average_sharpness=("sharpness_laplacian_var", "mean"),
    average_clip=("clip_similarity", "mean"),
    clip_evaluated=("clip_similarity", "count")
).reset_index()

print("\n" + "=" * 65)
print("MASTER EXPERIMENT SUMMARY")
print("=" * 65)

print(summary.round(4).to_string(index=False))

print("\nTotal experiment records:", len(master_df))
print("Master CSV:", master_output)

print("\nCLIP evaluation status:")
print(
    master_df["clip_status"]
    .fillna("not_evaluated")
    .value_counts()
    .to_string()
)
```

```python
# ============================================================
# CELL 24: FIX SCHEDULER NAMES AND REBUILD MASTER CSV
# ============================================================

import os
import pandas as pd
import numpy as np

experiment_files = [
    "baseline_fp32_pndm.csv",
    "fp16_pndm_30_512.csv",
    "fp16_ddim_30_512.csv",
    "fp16_pndm_step_optimization.csv",
    "fp16_ddim_step_optimization.csv",
    "fp16_pndm_30_384.csv",
    "combined_fp16_pndm_10_384.csv"
]

all_results = []

for filename in experiment_files:

    path = os.path.join(METRICS_DIR, filename)

    if os.path.exists(path):
        df = pd.read_csv(path)
        all_results.append(df)

# Combine
master_df = pd.concat(all_results, ignore_index=True)

# Normalize scheduler names
master_df["scheduler"] = (
    master_df["scheduler"]
    .astype(str)
    .str.replace("Scheduler", "", regex=False)
)

# Normalize precision
master_df["precision"] = (
    master_df["precision"].astype(str).str.upper()
)

# Load CLIP results
clip_df = pd.read_csv(
    os.path.join(
        METRICS_DIR,
        "clip_semantic_similarity.csv"
    )
)

clip_mapping = {
    "baseline_fp32_pndm_30_512": ("FP32", "PNDM", 30, 512),
    "fp16_pndm_30_512": ("FP16", "PNDM", 30, 512),
    "fp16_ddim_30_512": ("FP16", "DDIM", 30, 512),
    "fp16_pndm_30_384": ("FP16", "PNDM", 30, 384),
    "combined_fp16_pndm_10_384": ("FP16", "PNDM", 10, 384)
}

for column, position in [
    ("precision", 0),
    ("scheduler", 1),
    ("steps", 2),
    ("height", 3),
    ("width", 3)
]:
    clip_df[column] = clip_df["configuration"].map(
        lambda x: clip_mapping.get(x, (None,)*4)[position]
    )

clip_df = clip_df.rename(
    columns={"status": "clip_status"}
)

merge_columns = [
    "precision",
    "scheduler",
    "steps",
    "height",
    "width",
    "prompt_id"
]

clip_merge = clip_df[
    merge_columns + ["clip_similarity", "clip_status"]
]

# Remove any old CLIP columns before merging
master_df = master_df.drop(
    columns=["clip_similarity", "clip_status"],
    errors="ignore"
)

master_df = master_df.merge(
    clip_merge,
    on=merge_columns,
    how="left"
)

# Identify configurations
def get_config(row):

    return (
        f"{row['precision']}_"
        f"{row['scheduler']}_"
        f"{int(row['steps'])}_"
        f"{int(row['height'])}"
    )

master_df["configuration"] = master_df.apply(
    get_config,
    axis=1
)

# Save corrected master dataset
master_output = os.path.join(
    METRICS_DIR,
    "M1_SD15_MASTER_RESULTS.csv"
)

master_df.to_csv(master_output, index=False)

# Summary
summary = master_df.groupby(
    "configuration"
).agg(
    image_count=("prompt_id", "count"),
    average_latency=("latency_seconds", "mean"),
    average_allocated_vram=("peak_allocated_mb", "mean"),
    average_reserved_vram=("peak_reserved_mb", "mean"),
    average_sharpness=("sharpness_laplacian_var", "mean"),
    average_clip=("clip_similarity", "mean"),
    clip_evaluated=("clip_similarity", "count")
).reset_index()

print("=" * 70)
print("CORRECTED MASTER EXPERIMENT SUMMARY")
print("=" * 70)

print(summary.round(5).to_string(index=False))

print("\nCLIP evaluation status:")
print(
    master_df["clip_status"]
    .fillna("not_evaluated")
    .value_counts()
)

print("\nTotal experiment records:", len(master_df))
print("Saved:", master_output)

# Validation
assert len(master_df) == 132
assert summary["clip_evaluated"].sum() == 59

print("\nVALIDATION PASSED")
```

```python
# ============================================================
# CELL 28: MEASURE ACTUAL LOADED PIPELINE SIZE
# ============================================================

import torch
import json
import os

print("=" * 65)
print("ACTUAL LOADED PIPELINE FOOTPRINT")
print("=" * 65)

component_sizes = {}
total_parameter_bytes = 0
total_buffer_bytes = 0

for component_name, component in pipe.components.items():

    if component is None:
        continue

    if not isinstance(component, torch.nn.Module):
        continue

    parameter_bytes = sum(
        p.numel() * p.element_size()
        for p in component.parameters()
    )

    buffer_bytes = sum(
        b.numel() * b.element_size()
        for b in component.buffers()
    )

    component_bytes = parameter_bytes + buffer_bytes

    total_parameter_bytes += parameter_bytes
    total_buffer_bytes += buffer_bytes

    component_sizes[component_name] = {
        "parameter_mb": round(parameter_bytes / (1024 ** 2), 2),
        "buffer_mb": round(buffer_bytes / (1024 ** 2), 2),
        "total_mb": round(component_bytes / (1024 ** 2), 2)
    }

    print(
        f"{component_name:<20}"
        f"{component_bytes / (1024 ** 2):>12.2f} MB"
    )

total_loaded_bytes = total_parameter_bytes + total_buffer_bytes

print("-" * 65)
print(
    f"{'TOTAL':<20}"
    f"{total_loaded_bytes / (1024 ** 2):>12.2f} MB"
)

print(
    f"\nParameter storage: "
    f"{total_parameter_bytes / (1024 ** 3):.3f} GB"
)

print(
    f"Buffer storage: "
    f"{total_buffer_bytes / (1024 ** 3):.3f} GB"
)

print(
    f"Total loaded component storage: "
    f"{total_loaded_bytes / (1024 ** 3):.3f} GB"
)

# Update metadata
metadata_path = os.path.join(
    METADATA_DIR,
    "M1_SD15_environment_metadata.json"
)

with open(metadata_path, "r") as f:
    metadata = json.load(f)

metadata["model"]["loaded_pipeline_component_sizes"] = component_sizes

metadata["model"]["loaded_pipeline_parameter_size_mb"] = round(
    total_parameter_bytes / (1024 ** 2), 2
)

metadata["model"]["loaded_pipeline_buffer_size_mb"] = round(
    total_buffer_bytes / (1024 ** 2), 2
)

metadata["model"]["loaded_pipeline_total_size_mb"] = round(
    total_loaded_bytes / (1024 ** 2), 2
)

with open(metadata_path, "w") as f:
    json.dump(metadata, f, indent=4)

print("\nMetadata updated successfully.")
```

```python
def select_generation_profile(memory_budget_gb, latency_budget_s):

    # Try profiles from highest to lowest quality preference
    priority = ["QUALITY", "BALANCED", "LOW"]

    for profile_name in priority:

        config = GENERATION_PROFILES[profile_name]

        memory_required = config["expected_memory_gb"]
        latency_required = config["expected_latency"]

        # Check whether the configuration fits both budgets
        if (
            memory_required <= memory_budget_gb
            and latency_required <= latency_budget_s
        ):

            return {
                "selected_profile": profile_name,
                "configuration": config,
                "status": "Selected successfully"
            }

    # No tested configuration satisfies both constraints
    return {
        "selected_profile": None,
        "configuration": None,
        "status": "Insufficient resource budget"
    }


# Test different resource budgets
test_budgets = [
    {"memory": 3.1, "latency": 0.5},
    {"memory": 3.1, "latency": 1.0},
    {"memory": 3.3, "latency": 5.0},
    {"memory": 2.5, "latency": 1.0}
]

for budget in test_budgets:

    result = select_generation_profile(
        memory_budget_gb=budget["memory"],
        latency_budget_s=budget["latency"]
    )

    print("\nResource Budget:", budget)
    print("Selected Profile:", result["selected_profile"])
    print("Status:", result["status"])
```

```python
def generate_with_fallback(
    prompt,
    memory_budget_gb,
    latency_budget_s,
    generation_fn
):

    # Determine the preferred profile
    initial_result = select_generation_profile(
        memory_budget_gb,
        latency_budget_s
    )

    if initial_result["selected_profile"] is None:
        return {
            "status": "Failed",
            "reason": "No configuration fits the resource budget",
            "selected_profile": None,
            "image": None
        }

    preferred = initial_result["selected_profile"]

    # Try preferred profile first, then progressively lighter profiles
    priority = ["QUALITY", "BALANCED", "LOW"]
    start_index = priority.index(preferred)
    candidates = priority[start_index:]

    failed_attempts = []

    for profile_name in candidates:

        config = GENERATION_PROFILES[profile_name]

        # Verify resource constraints
        if (
            config["expected_memory_gb"] > memory_budget_gb
            or config["expected_latency"] > latency_budget_s
        ):
            continue

        try:
            print(f"Trying {profile_name} profile...")

            image = generation_fn(
                prompt,
                profile_name,
                config
            )

            return {
                "status": "Success",
                "selected_profile": profile_name,
                "image": image,
                "failed_attempts": failed_attempts
            }

        except Exception as error:

            print(f"{profile_name} failed: {error}")

            failed_attempts.append({
                "profile": profile_name,
                "error": str(error)
            })

    return {
        "status": "Failed",
        "reason": "All feasible configurations failed",
        "selected_profile": None,
        "image": None,
        "failed_attempts": failed_attempts
    }
```

```python
import torch

def complete_adaptive_generation(
    prompt,
    memory_budget_gb,
    latency_budget_s
):

    selection = select_generation_profile(
        memory_budget_gb,
        latency_budget_s
    )

    if selection["selected_profile"] is None:
        return {
            "status": "Failure",
            "message": "No profile satisfies the estimated resource budget.",
            "failed_attempts": []
        }

    priority = ["QUALITY", "BALANCED", "LOW"]

    start_index = priority.index(
        selection["selected_profile"]
    )

    candidates = priority[start_index:]
    failed_attempts = []

    for profile_name in candidates:

        config = GENERATION_PROFILES[profile_name]

        if config["expected_memory_gb"] > memory_budget_gb:
            continue

        if config["expected_latency"] > latency_budget_s:
            continue

        try:
            print(f"\nTrying {profile_name} profile...")

            load_required_model(config["model"])

            if config["model"] == "Stable Diffusion Turbo":
                pipe = turbo_pipe
                generation_fn = real_turbo_generation
            else:
                pipe = sd15_pipe
                generation_fn = real_sd15_generation

            if pipe.safety_checker is None:
                raise RuntimeError("Safety checker is not enabled.")

            generation = generation_fn(
                prompt,
                profile_name,
                config
            )

            actual_latency = generation["latency"]
            actual_memory = generation["peak_memory_gb"]

            violations = []

            if actual_latency > latency_budget_s:
                violations.append("Latency budget exceeded")

            if actual_memory > memory_budget_gb:
                violations.append("Memory budget exceeded")

            if violations:
                failed_attempts.append({
                    "profile": profile_name,
                    "violations": violations,
                    "actual_latency": actual_latency,
                    "actual_memory_gb": actual_memory
                })

                print("Resource validation failed:", violations)
                continue

            return {
                "status": "Success",
                "selected_profile": profile_name,
                "configuration": config,
                "image": generation["image"],
                "actual_latency": actual_latency,
                "actual_memory_gb": actual_memory,
                "resource_validation": "Passed",
                "failed_attempts": failed_attempts
            }

        except torch.cuda.OutOfMemoryError as error:

            failed_attempts.append({
                "profile": profile_name,
                "error": str(error)
            })

            print("GPU memory exhausted.")

            unload_current_models()

        except Exception as error:

            failed_attempts.append({
                "profile": profile_name,
                "error": str(error)
            })

            print("Generation failed:", error)

            unload_current_models()

    return {
        "status": "Failure",
        "message": "No profile completed within the resource budgets.",
        "failed_attempts": failed_attempts
    }

print("Validated adaptive controller created!")
```

```python
def load_required_model(model_name):

    global turbo_pipe, sd15_pipe

    # Reuse a model only if its safety checker is enabled
    if model_name == "Stable Diffusion Turbo":

        if (
            "turbo_pipe" in globals()
            and turbo_pipe is not None
            and turbo_pipe.safety_checker is not None
        ):
            print("Safety-enabled SD-Turbo already loaded.")
            return

    elif model_name == "Stable Diffusion 1.5":

        if (
            "sd15_pipe" in globals()
            and sd15_pipe is not None
            and sd15_pipe.safety_checker is not None
        ):
            print("Safety-enabled SD1.5 already loaded.")
            return

    else:
        raise ValueError("Unknown model: " + model_name)

    # Unload previous model
    unload_current_models()

    if model_name == "Stable Diffusion Turbo":

        turbo_pipe = StableDiffusionPipeline.from_pretrained(
            "stabilityai/sd-turbo",
            torch_dtype=torch.float16,
            use_safetensors=True,
            safety_checker=safety_checker,
            feature_extractor=feature_extractor,
            requires_safety_checker=True
        )

        turbo_pipe.scheduler = EulerDiscreteScheduler.from_config(
            turbo_pipe.scheduler.config
        )

        turbo_pipe = turbo_pipe.to("cuda")
        turbo_pipe.set_progress_bar_config(disable=True)

        print("SD-Turbo loaded with safety checker!")

    else:

        sd15_pipe = StableDiffusionPipeline.from_pretrained(
            "stable-diffusion-v1-5/stable-diffusion-v1-5",
            torch_dtype=torch.float16,
            use_safetensors=True,
            safety_checker=safety_checker,
            feature_extractor=feature_extractor,
            requires_safety_checker=True
        )

        sd15_pipe.scheduler = DDIMScheduler.from_config(
            sd15_pipe.scheduler.config
        )

        sd15_pipe = sd15_pipe.to("cuda")
        sd15_pipe.set_progress_bar_config(disable=True)

        print("SD1.5 loaded with safety checker!")

    # Verify after loading
    active_pipe = (
        turbo_pipe
        if model_name == "Stable Diffusion Turbo"
        else sd15_pipe
    )

    assert active_pipe.safety_checker is not None
    assert active_pipe.feature_extractor is not None

    print("Safety checker verified:", True)
    print("Feature extractor verified:", True)
```

```python
from skimage.metrics import (
    peak_signal_noise_ratio,
    structural_similarity
)

# Upscale reduced images to original resolution
image_384_up = cv2.resize(
    image_384,
    (512, 512),
    interpolation=cv2.INTER_CUBIC
)

image_256_up = cv2.resize(
    image_256,
    (512, 512),
    interpolation=cv2.INTER_CUBIC
)

def calculate_quality(reference, candidate):

    psnr = peak_signal_noise_ratio(
        reference,
        candidate,
        data_range=255
    )

    ssim = structural_similarity(
        reference,
        candidate,
        channel_axis=2,
        data_range=255
    )

    sharpness = cv2.Laplacian(
        cv2.cvtColor(candidate, cv2.COLOR_RGB2GRAY),
        cv2.CV_64F
    ).var()

    return psnr, ssim, sharpness


quality_results = {
    "Original 512x512": calculate_quality(
        original_image, original_image
    ),
    "384x384": calculate_quality(
        original_image, image_384_up
    ),
    "256x256": calculate_quality(
        original_image, image_256_up
    )
}

for name, values in quality_results.items():

    print(f"\n{name}")
    print(f"PSNR: {values[0]:.3f} dB")
    print(f"SSIM: {values[1]:.4f}")
    print(f"Sharpness: {values[2]:.2f}")
```

```python
# Convert original image to grayscale
gray_original = cv2.cvtColor(
    original_image,
    cv2.COLOR_RGB2GRAY
)

# Apply median filtering
median_image = cv2.medianBlur(
    gray_original,
    3
)

# Calculate absolute difference
difference = cv2.absdiff(
    gray_original,
    median_image
)

# Detect unusually large local differences
threshold_value = 35

_, artifact_mask = cv2.threshold(
    difference,
    threshold_value,
    255,
    cv2.THRESH_BINARY
)

# Count detected candidate pixels
candidate_pixels = np.count_nonzero(artifact_mask)
total_pixels = artifact_mask.size

print("ARTIFACT DETECTION RESULTS")
print("-" * 40)
print("Image dimensions:", gray_original.shape)
print("Candidate pixels:", candidate_pixels)
print("Total pixels:", total_pixels)
print(
    "Candidate percentage:",
    round(candidate_pixels / total_pixels * 100, 3),
    "%"
)

# Visualize results
fig, axes = plt.subplots(1, 3, figsize=(16, 5))

axes[0].imshow(gray_original, cmap="gray")
axes[0].set_title("Original Grayscale")

axes[1].imshow(difference, cmap="hot")
axes[1].set_title("Local Difference Map")

axes[2].imshow(artifact_mask, cmap="gray")
axes[2].set_title("Potential Artifact Candidates")

for ax in axes:
    ax.axis("off")

plt.tight_layout()
plt.show()
```

```python

# STEP 3: Edge-aware artifact mask refinement

# Detect strong edges in the original grayscale image
edges = cv2.Canny(
    gray_original,
    threshold1=80,
    threshold2=160
)

# Slightly expand the detected edges to protect nearby pixels
kernel = np.ones((3, 3), np.uint8)

protected_edges = cv2.dilate(
    edges,
    kernel,
    iterations=1
)

# Remove protected edge pixels from the artifact candidates
refined_artifact_mask = cv2.bitwise_and(
    artifact_mask,
    cv2.bitwise_not(protected_edges)
)

# Calculate refined candidate count
original_candidates = np.count_nonzero(artifact_mask)
refined_candidates = np.count_nonzero(refined_artifact_mask)

print("EDGE-AWARE MASK REFINEMENT")
print("-" * 40)
print("Original candidates:", original_candidates)
print("Refined candidates:", refined_candidates)
print(
    "Removed from candidate mask:",
    original_candidates - refined_candidates
)
print(
    "Remaining candidate percentage:",
    round(refined_candidates / refined_artifact_mask.size * 100, 3),
    "%"
)

# Visualize the results
fig, axes = plt.subplots(1, 3, figsize=(16, 5))

axes[0].imshow(edges, cmap="gray")
axes[0].set_title("Detected Natural Edges")

axes[1].imshow(artifact_mask, cmap="gray")
axes[1].set_title("Original Artifact Mask")

axes[2].imshow(refined_artifact_mask, cmap="gray")
axes[2].set_title("Edge-Protected Artifact Mask")

for ax in axes:
    ax.axis("off")

plt.tight_layout()
plt.show()
```

```python

def process_generated_image(
    image,
    apply_clahe=True,
    apply_bilateral=False,
    apply_artifact_correction=False
):
    """
    Apply optional classical CV processing to a generated PIL image.
    The original input image remains unchanged.
    """

    # Convert PIL image into RGB NumPy array
    input_array = np.array(image.convert("RGB"))

    # Work on a separate copy
    processed = input_array.copy()

    # CLAHE enhancement in LAB colour space
    if apply_clahe:
        lab = cv2.cvtColor(
            processed,
            cv2.COLOR_RGB2LAB
        )

        L, A, B = cv2.split(lab)

        clahe = cv2.createCLAHE(
            clipLimit=1.0,
            tileGridSize=(8, 8)
        )

        enhanced_L = clahe.apply(L)

        enhanced_lab = cv2.merge([
            enhanced_L, A, B
        ])

        processed = cv2.cvtColor(
            enhanced_lab,
            cv2.COLOR_LAB2RGB
        )

    # Optional edge-preserving smoothing
    if apply_bilateral:
        processed = cv2.bilateralFilter(
            processed,
            d=9,
            sigmaColor=75,
            sigmaSpace=75
        )

    # Optional conservative artifact correction
    if apply_artifact_correction:
        gray = cv2.cvtColor(
            processed,
            cv2.COLOR_RGB2GRAY
        )

        median = cv2.medianBlur(gray, 3)

        difference = cv2.absdiff(
            gray, median
        )

        _, mask = cv2.threshold(
            difference,
            35,
            255,
            cv2.THRESH_BINARY
        )

        edges = cv2.Canny(gray, 80, 160)

        protected_edges = cv2.dilate(
            edges,
            np.ones((3, 3), np.uint8),
            iterations=1
        )

        refined_mask = cv2.bitwise_and(
            mask,
            cv2.bitwise_not(protected_edges)
        )

        processed = cv2.inpaint(
            processed,
            refined_mask,
            3,
            cv2.INPAINT_TELEA
        )

    # Return PIL image
    return Image.fromarray(processed)


print("CV processing function created successfully!")
```

```python

def complete_adaptive_generation(
    prompt,
    memory_budget_gb,
    latency_budget_s,
    apply_clahe=True,
    apply_bilateral=False,
    apply_artifact_correction=False
):
    # Step 1: Select the preferred generation profile
    selection = select_generation_profile(
        memory_budget_gb,
        latency_budget_s
    )

    if selection["selected_profile"] is None:
        return {
            "status": "Failure",
            "message": "No profile satisfies the estimated resource budget.",
            "failed_attempts": []
        }

    priority = ["QUALITY", "BALANCED", "LOW"]

    start_index = priority.index(
        selection["selected_profile"]
    )

    candidates = priority[start_index:]
    failed_attempts = []

    # Step 2: Try the selected profile and lighter alternatives
    for profile_name in candidates:

        config = GENERATION_PROFILES[profile_name]

        if config["expected_memory_gb"] > memory_budget_gb:
            continue

        if config["expected_latency"] > latency_budget_s:
            continue

        try:
            print(f"\nTrying {profile_name} profile...")

            # Load the required model
            load_required_model(config["model"])

            if config["model"] == "Stable Diffusion Turbo":
                pipe = turbo_pipe
                generation_fn = real_turbo_generation
            else:
                pipe = sd15_pipe
                generation_fn = real_sd15_generation

            # Safety checker validation
            if pipe.safety_checker is None:
                raise RuntimeError("Safety checker is not enabled.")

            # Step 3: Generate original image
            generation = generation_fn(
                prompt,
                profile_name,
                config
            )

            actual_latency = generation["latency"]
            actual_memory = generation["peak_memory_gb"]

            # Step 4: Validate actual resource consumption
            violations = []

            if actual_latency > latency_budget_s:
                violations.append("Latency budget exceeded")

            if actual_memory > memory_budget_gb:
                violations.append("Memory budget exceeded")

            if violations:
                failed_attempts.append({
                    "profile": profile_name,
                    "violations": violations,
                    "actual_latency": actual_latency,
                    "actual_memory_gb": actual_memory
                })

                print("Resource validation failed:", violations)
                continue

            # Step 5: Preserve original image
            original_image = generation["image"]

            # Step 6: Apply optional classical CV processing
            cv_start = time.perf_counter()

            try:
                processed_image = process_generated_image(
                    original_image,
                    apply_clahe=apply_clahe,
                    apply_bilateral=apply_bilateral,
                    apply_artifact_correction=apply_artifact_correction
                )

                cv_status = "Applied"

            except Exception as cv_error:
                print("CV processing failed:", cv_error)

                # Preserve the generated image if CV fails
                processed_image = original_image.copy()
                cv_status = "Skipped due to error"

            cv_latency = time.perf_counter() - cv_start

            # Step 7: Return complete generation result
            return {
                "status": "Success",
                "selected_profile": profile_name,
                "configuration": config,

                "original_image": original_image,
                "processed_image": processed_image,

                "actual_latency": actual_latency,
                "actual_memory_gb": actual_memory,

                "cv_latency": cv_latency,
                "total_pipeline_latency": actual_latency + cv_latency,
                "cv_status": cv_status,

                "resource_validation": "Passed",
                "failed_attempts": failed_attempts
            }

        except torch.cuda.OutOfMemoryError as error:

            failed_attempts.append({
                "profile": profile_name,
                "error": str(error)
            })

            print("GPU memory exhausted.")
            unload_current_models()

        except Exception as error:

            failed_attempts.append({
                "profile": profile_name,
                "error": str(error)
            })

            print("Generation failed:", error)
            unload_current_models()

    return {
        "status": "Failure",
        "message": "No profile completed within the resource budgets.",
        "failed_attempts": failed_attempts
    }


print("Integrated adaptive generation controller created successfully!")
```

```python

# Initialize CLIP model for semantic evaluation

import torch
from transformers import CLIPModel, CLIPProcessor

CLIP_MODEL_NAME = "openai/clip-vit-base-patch32"

CLIP_DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

print("Loading CLIP model...")

clip_model = CLIPModel.from_pretrained(
    CLIP_MODEL_NAME
).to(CLIP_DEVICE)

clip_processor = CLIPProcessor.from_pretrained(
    CLIP_MODEL_NAME
)

clip_model.eval()

print("\nCLIP MODEL INITIALIZED")
print("-" * 45)
print("Model:", CLIP_MODEL_NAME)
print("Device:", CLIP_DEVICE)
print("Model loaded:", clip_model is not None)
print("Processor loaded:", clip_processor is not None)
```

```python

import torch
import torch.nn.functional as F
import numpy as np

def calculate_clip_similarity(image, prompt):

    # Prepare image and text inputs
    inputs = clip_processor(
        text=[prompt],
        images=image.convert("RGB"),
        return_tensors="pt",
        padding=True
    )

    # Move inputs to GPU
    inputs = {
        key: value.to(CLIP_DEVICE)
        for key, value in inputs.items()
    }

    # Extract embeddings without calculating gradients
    with torch.no_grad():

        outputs = clip_model(**inputs)

        image_embedding = outputs.image_embeds
        text_embedding = outputs.text_embeds

        # Normalize embeddings
        image_embedding = F.normalize(
            image_embedding,
            p=2,
            dim=-1
        )

        text_embedding = F.normalize(
            text_embedding,
            p=2,
            dim=-1
        )

        # Cosine similarity
        similarity = torch.sum(
            image_embedding * text_embedding,
            dim=-1
        )

    return float(similarity.item())


print("CLIP semantic evaluation function created successfully!")
```

```python

import gradio as gr
import torch
import gc
import time

def litegen_generate(
    prompt,
    mode,
    memory_budget,
    latency_budget,
    apply_clahe,
    apply_bilateral,
    apply_artifact_correction
):
    if not prompt or not prompt.strip():
        raise gr.Error("Please enter a text prompt.")

    # Map the selected UI mode to a suitable resource budget.
    if mode == "Speed":
        memory_budget = min(memory_budget, 3.7)
        latency_budget = min(latency_budget, 1.0)

    elif mode == "Balanced":
        memory_budget = min(memory_budget, 3.7)
        latency_budget = min(latency_budget, 1.5)

    elif mode == "Quality":
        latency_budget = max(latency_budget, 6.0)

    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    try:
        result = complete_adaptive_generation(
            prompt=prompt.strip(),
            memory_budget_gb=memory_budget,
            latency_budget_s=latency_budget,
            apply_clahe=apply_clahe,
            apply_bilateral=apply_bilateral,
            apply_artifact_correction=apply_artifact_correction
        )

        if result.get("status") != "Success":
            raise gr.Error(
                result.get("message", "Image generation failed.")
            )

        original = result["original_image"]
        processed = result["processed_image"]

        report = f"""
### Generation completed

| Metric | Result |
|---|---:|
| Selected profile | {result['selected_profile']} |
| Inference latency | {result['actual_latency']:.3f} seconds |
| CV processing latency | {result['cv_latency']:.4f} seconds |
| Total pipeline latency | {result['total_pipeline_latency']:.3f} seconds |
| Peak GPU memory | {result['actual_memory_gb']:.3f} GB |
| CV status | {result['cv_status']} |
| Resource validation | {result['resource_validation']} |
"""

        return original, processed, report

    except gr.Error:
        raise

    except Exception as e:
        raise gr.Error(f"Generation error: {str(e)}")


with gr.Blocks(title="LiteGen AI") as litegen_app:

    gr.Markdown(
        """
        # LiteGen AI
        ### Resource-Efficient Text-to-Image Generation

        Generate images using adaptive diffusion inference,
        resource-aware profile selection and optional classical
        computer vision enhancement.
        """
    )

    with gr.Row():

        with gr.Column(scale=1):

            prompt_input = gr.Textbox(
                label="Enter your image prompt",
                placeholder="Describe the image you want to generate...",
                lines=4
            )

            mode_input = gr.Radio(
                choices=["Speed", "Balanced", "Quality"],
                value="Balanced",
                label="Generation Mode"
            )

            memory_input = gr.Slider(
                minimum=2.0,
                maximum=6.0,
                value=3.7,
                step=0.1,
                label="GPU Memory Budget (GB)"
            )

            latency_input = gr.Slider(
                minimum=0.2,
                maximum=10.0,
                value=1.0,
                step=0.1,
                label="Latency Budget (seconds)"
            )

            gr.Markdown("### Computer Vision Enhancements")

            clahe_input = gr.Checkbox(
                value=False,
                label="Apply CLAHE contrast enhancement"
            )

            bilateral_input = gr.Checkbox(
                value=False,
                label="Apply bilateral filtering"
            )

            artifact_input = gr.Checkbox(
                value=False,
                label="Apply artifact correction"
            )

            generate_button = gr.Button(
                "Generate Image",
                variant="primary"
            )

        with gr.Column(scale=1):

            original_output = gr.Image(
                label="Original Generated Image",
                type="pil"
            )

            processed_output = gr.Image(
                label="Processed Image",
                type="pil"
            )

    metrics_output = gr.Markdown(
        label="Generation Performance"
    )

    generate_button.click(
        fn=litegen_generate,
        inputs=[
            prompt_input,
            mode_input,
            memory_input,
            latency_input,
            clahe_input,
            bilateral_input,
            artifact_input
        ],
        outputs=[
            original_output,
            processed_output,
            metrics_output
        ]
    )

print("LiteGen AI interface created successfully.")
```

```python

DEFAULT_NEGATIVE_PROMPT = (
    "deformed, distorted, malformed objects, broken geometry, "
    "duplicate objects, unnatural proportions, blurry, "
    "low quality, visual artifacts"
)

def real_sd15_generation(prompt, profile_name, config):
    if config["model"] != "Stable Diffusion 1.5":
        raise ValueError("Incorrect model configuration for SD1.5.")

    generator = torch.Generator(device=DEVICE).manual_seed(12345)

    torch.cuda.synchronize()
    torch.cuda.reset_peak_memory_stats()
    start_time = time.perf_counter()

    result = sd15_pipe(
        prompt=prompt,
        negative_prompt=DEFAULT_NEGATIVE_PROMPT,
        num_inference_steps=config["steps"],
        guidance_scale=config["guidance_scale"],
        height=512,
        width=512,
        generator=generator
    )

    torch.cuda.synchronize()

    latency = time.perf_counter() - start_time
    peak_memory = torch.cuda.max_memory_allocated() / (1024**3)

    return {
        "image": result.images[0],
        "profile": profile_name,
        "latency": latency,
        "peak_memory_gb": peak_memory
    }

print("SD1.5 generation function updated.")
print("Negative prompting: enabled")
print("Existing Gradio function arguments: preserved")
```

```python

import os
import shutil
import pandas as pd

COMBINED_SOURCE = (
    "/kaggle/input/datasets/shamantnagarajm/"
    "litegen-ai-combined-model-results-csv/"
    "LiteGen_AI_Combined_Model_Results.csv"
)

PROJECT_DIR = "/kaggle/working/Recovered_Experiments"

os.makedirs(PROJECT_DIR, exist_ok=True)

COMBINED_DEST = os.path.join(
    PROJECT_DIR,
    "LiteGen_AI_Combined_Model_Results.csv"
)

shutil.copy2(COMBINED_SOURCE, COMBINED_DEST)

combined_df = pd.read_csv(COMBINED_DEST)

print("Combined benchmark restored successfully.")
print("Rows:", combined_df.shape[0])
print("Columns:", combined_df.shape[1])

print("\nAvailable models/configurations:")
display(combined_df.head())

print("\nMissing values:")
display(combined_df.isnull().sum().to_frame("Missing Values"))
```

```python

print("MODEL DISTRIBUTION")
display(
    combined_df.groupby("model").agg(
        rows=("prompt_id", "count"),
        unique_prompts=("prompt_id", "nunique"),
        configurations=("run_id", "nunique"),
        mean_latency=("latency_mean_seconds", "mean"),
        mean_clip=("clip_score", "mean")
    ).round(4)
)

print("\nCONFIGURATION DISTRIBUTION")
display(
    combined_df.groupby(
        ["model", "run_id", "precision", "scheduler", "steps", "resolution"],
        dropna=False
    ).agg(
        samples=("prompt_id", "count"),
        mean_latency=("latency_mean_seconds", "mean"),
        mean_clip=("clip_score", "mean"),
        mean_memory=("peak_allocated_gb", "mean")
    ).round(4).reset_index()
)

print("\nVALIDATION")
print("Total rows:", len(combined_df))
print("Unique prompts:", combined_df["prompt_id"].nunique())
print("Unique models:", combined_df["model"].nunique())
print("Duplicate rows:", combined_df.duplicated().sum())
```

```python

config_columns = [
    "model",
    "precision",
    "scheduler",
    "steps",
    "guidance_scale",
    "height",
    "width"
]

configuration_summary = (
    combined_df
    .groupby(config_columns, dropna=False)
    .agg(
        sample_count=("prompt_id", "count"),
        unique_prompts=("prompt_id", "nunique"),
        mean_latency=("latency_mean_seconds", "mean"),
        mean_allocated_memory=("peak_allocated_gb", "mean"),
        mean_reserved_memory=("peak_reserved_gb", "mean"),
        mean_clip=("clip_score", "mean"),
        mean_sharpness=("sharpness", "mean")
    )
    .reset_index()
)

configuration_summary["resolution"] = (
    configuration_summary["width"].astype(str)
    + "x"
    + configuration_summary["height"].astype(str)
)

configuration_summary = configuration_summary.drop(
    columns=["height", "width"]
)

configuration_summary = configuration_summary.round(4)

print("CORRECTED CONFIGURATION SUMMARY")
print("Distinct configurations:", len(configuration_summary))
display(configuration_summary)

print("\nSAVED CONFIGURATION SUMMARY")
summary_path = os.path.join(
    PROJECT_DIR,
    "LiteGen_AI_Configuration_Summary.csv"
)

configuration_summary.to_csv(summary_path, index=False)
print(summary_path)
```

```python

import os
import zipfile

FINAL = "/kaggle/working/LiteGen_AI_Final_Submission"
ZIP_PATH = "/kaggle/working/LiteGen_AI_Final_Submission.zip"

if os.path.exists(ZIP_PATH):
    os.remove(ZIP_PATH)

with zipfile.ZipFile(
    ZIP_PATH, "w", compression=zipfile.ZIP_DEFLATED,
    compresslevel=6
) as zipf:

    for root, dirs, files in os.walk(FINAL):
        for filename in files:
            full_path = os.path.join(root, filename)
            relative_path = os.path.relpath(
                full_path,
                os.path.dirname(FINAL)
            )
            zipf.write(full_path, relative_path)

print("FINAL ZIP CREATED")
print("=" * 45)
print("Archive:", ZIP_PATH)
print("Size:", round(os.path.getsize(ZIP_PATH) / (1024 ** 2), 2), "MB")

with zipfile.ZipFile(ZIP_PATH, "r") as zipf:
    print("Files in archive:", len(zipf.namelist()))
    print("Archive integrity:", "PASSED" if zipf.testzip() is None else "FAILED")
```
